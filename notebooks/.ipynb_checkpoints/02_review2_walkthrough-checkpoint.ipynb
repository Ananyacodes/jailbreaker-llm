# 02 · Review 2 walkthrough: finding compliance states without labels

PCA, K-means, Gaussian mixtures (EM), hierarchical clustering and a hidden Markov model, run cell by cell on red-team judge scores.

**The question:** can the hidden compliance state of each turn be recovered from the judge's three probabilities alone, with no labels?
**How it is scored without cheating:** the true state is touched only in the scoring cells (matched accuracy, ARI). Silhouette needs no labels at all.

Every algorithm here comes from `src/analysis/unsupervised.py` (NumPy, from the equations) or `src/analysis/trace.py` (the same algorithms recording every step so we can look inside). `tests/test_trace.py` proves both give identical answers. The one exception is the HMM, which uses the `hmmlearn` library.

Choose a campaign in the next cell and re-run everything below.

````python
import os, sys, warnings
warnings.filterwarnings("ignore")
if os.path.basename(os.getcwd()) == "notebooks":   # run from the repo root so relative data paths work
    os.chdir("..")
sys.path.insert(0, os.getcwd())

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
plt.rcParams.update({"figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False})

from src.analysis import scenarios, trace, external_data as ext
from src.analysis.unsupervised import SCORE_COLUMNS, conversation_features, adjusted_rand
get_ipython().run_line_magic('matplotlib', 'inline')   # the repo's analysis modules force the Agg backend; switch back so plots show here
from src.analysis.utils import COMPLIANCE_STATES, load_transcript_rows

STATE_COLORS = ["#2F6FED", "#F2A20C", "#E5484D"]   # refusal, partial, full violation
PALETTE = ["#2F6FED", "#F2A20C", "#E5484D", "#16A34A", "#7C3AED", "#0EA5A4"]
````

````python
CAMPAIGN = "baseline"      # "baseline" | "noisy" | "stealthy" | "persistent"
SEED = 42
K = 3

def load_campaign(name):
    if name == "baseline":
        try:
            return load_transcript_rows().dropna(subset=["true_compliance_state"])
        except FileNotFoundError:
            from data.generate_synthetic_campaign import generate
            import json
            os.makedirs("data/transcripts", exist_ok=True)
            with open("data/transcripts/synthetic_campaign_pbl.jsonl", "w") as f:
                for r in generate():
                    f.write(json.dumps(r) + "\n")
            return load_transcript_rows().dropna(subset=["true_compliance_state"])
    return scenarios.simulate(name)

df = load_campaign(CAMPAIGN)
X_raw = df[SCORE_COLUMNS].values
y = df["true_compliance_state"].map(COMPLIANCE_STATES.index).values     # answer key: used ONLY for scoring
print(f"{CAMPAIGN}: {len(df)} turns, {df.conversation_id.nunique()} conversations")
df[["conversation_id", "timestep_in_conversation"] + SCORE_COLUMNS].head(8).round(3)
````

## Step 1 · Standardize

$$z = \frac{x - \mu}{\sigma}$$

Each score is rescaled to mean 0 and standard deviation 1, so no feature dominates only because of its scale.

````python
Xs, mu, sd = trace.zscore(X_raw)
pd.DataFrame({"mean before": X_raw.mean(0), "std before": X_raw.std(0), "mean after": Xs.mean(0), "std after": Xs.std(0)}, index=SCORE_COLUMNS).round(4)
````

## Step 2 · PCA

$$S = \tfrac1N (X-\bar x)^T (X-\bar x), \qquad S\,u_i = \lambda_i u_i, \qquad z = U^T (x-\bar x)$$

The three probabilities always add up to 1, so the last eigenvalue should be (almost exactly) zero: the data really lives in two dimensions.

````python
pca = trace.pca_trace(Xs)
display(pd.DataFrame({"eigenvalue": pca["eigvals"], "share of variance": pca["ratio"], "cumulative": np.cumsum(pca["ratio"])}, index=["PC1", "PC2", "PC3"]).round(5))
Z = (Xs - pca["mean"]) @ pca["eigvecs"][:, :2]     # keep PC1 and PC2

fig, axes = plt.subplots(1, 2, figsize=(11, 3.6))
axes[0].bar(["PC1", "PC2", "PC3"], pca["ratio"], color="#0D9488"); axes[0].set_title("Scree plot (share of variance)")
axes[1].scatter(Z[:, 0], Z[:, 1], s=6, c="#94A3B8", alpha=.7); axes[1].set_title("Turns projected onto PC1 and PC2"); axes[1].set_xlabel("PC1"); axes[1].set_ylabel("PC2")
plt.tight_layout(); plt.show()
````

## Step 3 · K-means

$$J = \sum_n \lVert x_n - \mu_{k(n)} \rVert^2$$

Assign every point to its nearest centre, move each centre to the mean of its points, repeat until nothing moves. Ten random restarts from k-means++ starts; the lowest J wins.

````python
best, runs = trace.kmeans_trace(Z, K, seed=SEED)
hist = best["history"]
print(f"winning restart #{best['run'] + 1} of {len(runs)}, converged in {len(hist)} iterations, J = {best['J']:.3f}")
print("J of all restarts:", [round(r["J"], 1) for r in runs])

pick = sorted({0, 1, 2, len(hist)} & set(range(len(hist) + 1)))
fig, axes = plt.subplots(1, len(pick), figsize=(3.6 * len(pick), 3.2), sharex=True, sharey=True)
for ax, t in zip(np.atleast_1d(axes), pick):
    if t == 0:
        ax.scatter(Z[:, 0], Z[:, 1], s=5, c="#CBD5E1"); cen = best["init"]; ax.set_title("start: k-means++ centres")
    else:
        h = hist[t - 1]; ax.scatter(Z[:, 0], Z[:, 1], s=5, c=[PALETTE[l] for l in h["labels"]], alpha=.7); cen = h["centers_after"]; ax.set_title(f"iteration {t}")
    ax.scatter(cen[:, 0], cen[:, 1], s=170, marker="X", c="white", edgecolors="#0F172A", linewidths=1.6, zorder=5)
plt.tight_layout(); plt.show()

fig, ax = plt.subplots(figsize=(5, 2.6)); ax.plot([h["J"] for h in hist], "o-"); ax.set_xlabel("iteration"); ax.set_ylabel("J"); ax.set_title("Distortion never increases"); plt.show()
````

### Choosing k: elbow and silhouette

````python
from src.analysis.unsupervised import silhouette
sweep = []
for k in range(2, 7):
    b, _ = trace.kmeans_trace(Z, k, seed=SEED)
    sweep.append({"k": k, "distortion J": round(b["J"], 1), "silhouette": round(silhouette(Z, b["labels"]), 3)})
sweep = pd.DataFrame(sweep).set_index("k"); display(sweep)
fig, ax = plt.subplots(figsize=(5.4, 2.8)); ax.plot(sweep.index, sweep["distortion J"], "o-", color="#2F6FED"); ax.set_ylabel("J", color="#2F6FED")
ax2 = ax.twinx(); ax2.plot(sweep.index, sweep["silhouette"], "s-", color="#F2A20C"); ax2.set_ylabel("silhouette", color="#F2A20C"); ax.set_xlabel("k"); plt.show()
print("Lower J always wins with more clusters; the silhouette peak and the elbow are what to read.")
````

## Step 4 · Gaussian mixture fitted with EM

$$p(x)=\sum_k \pi_k\,\mathcal N(x\mid\mu_k,\Sigma_k), \qquad \gamma_{nk}=\frac{\pi_k\,\mathcal N(x_n\mid\mu_k,\Sigma_k)}{\sum_j \pi_j\,\mathcal N(x_n\mid\mu_j,\Sigma_j)}$$

E-step: each point gets a *responsibility* for every blob. M-step: refit each blob's weight, centre and covariance using those responsibilities. Started from the K-means labels.

````python
from matplotlib.patches import Ellipse
BASE_RGB = np.array([[int(c[i:i+2], 16) / 255 for i in (1, 3, 5)] for c in PALETTE])
gm = trace.gmm_trace(Z, K, seed=SEED)
gh = gm["history"]
print(f"EM ran {len(gh) - 1} iterations, final log-likelihood {gm['lls'][-1]:.1f}, BIC {gm['bic']:.1f}")

def ell(ax, m, c):
    w, v = np.linalg.eigh(c); ang = np.degrees(np.arctan2(v[1, 1], v[0, 1]))
    for s_, a in ((1, 1.0), (2, .45)):
        ax.add_patch(Ellipse(m, 2 * s_ * np.sqrt(w[1]), 2 * s_ * np.sqrt(w[0]), angle=ang, fill=False, ec="#0F172A", lw=1.6, alpha=a))

its = sorted({0, 1, 3, len(gh) - 1})
fig, axes = plt.subplots(1, len(its), figsize=(3.7 * len(its), 3.3), sharex=True, sharey=True)
for ax, t in zip(axes, its):
    h = gh[t]; ax.scatter(Z[:, 0], Z[:, 1], s=6, c=np.clip(h["R"] @ BASE_RGB[:K], 0, 1), alpha=.85)
    for j in range(K): ell(ax, h["mu"][j][:2], h["cov"][j][:2, :2]); ax.scatter(*h["mu"][j][:2], s=110, marker="X", c="white", edgecolors="#0F172A", zorder=5)
    ax.set_title("init from K-means" if t == 0 else f"EM iteration {t}", fontsize=9)
plt.tight_layout(); plt.show()

fig, ax = plt.subplots(figsize=(5, 2.6)); ax.plot(range(1, len(gm["lls"]) + 1), gm["lls"], color="#16A34A"); ax.set_xlabel("iteration"); ax.set_ylabel("log-likelihood"); ax.set_title("Never decreases"); plt.show()
````

**How one point is assigned.** The E-step arithmetic for a single, deliberately ambiguous point (the one the final model is least sure about):

````python
h = gh[-1]
i = int(np.argmin(h["R"].max(1)))
log_pi, log_dens, log_joint, resp = trace.point_breakdown(Z[i], h["pi"], h["mu"], h["cov"])
pd.DataFrame({"log pi": log_pi, "log N(x | mu, Sigma)": log_dens, "log pi + log N": log_joint, "responsibility": resp}, index=[f"component {j}" for j in range(K)]).round(4)
````

## Step 5 · Hierarchical (agglomerative) clustering

Start with every item alone, merge the two closest clusters again and again (Lance-Williams update for the distance between clusters), then cut the tree. It is run on **one row per conversation** (6 summary features) so the tree stays readable.

Linkage choices: *single* = closest pair, *complete* = farthest pair, *average* = mean of all pairs, *ward* = smallest increase in spread.

````python
from scipy.cluster.hierarchy import dendrogram, fcluster, linkage as sp_linkage
C, final_state = conversation_features(df)
Cs, _, _ = trace.zscore(C)
merges, Zl = trace.agglo_trace(Cs, "complete")
N = len(Cs)
cut_h = (Zl[N - K - 1, 2] + Zl[N - K, 2]) / 2

fig, ax = plt.subplots(figsize=(8, 3.4))
dendrogram(Zl, ax=ax, truncate_mode="lastp", p=30, no_labels=True, color_threshold=cut_h, above_threshold_color="#94A3B8")
ax.axhline(cut_h, color="#E5484D", ls="--"); ax.set_ylabel("merge distance"); ax.set_title(f"Dendrogram of {N} conversations (last 30 merges), cut for k = {K}")
plt.show()

rows = []
for link in ("single", "complete", "average", "ward"):
    mg, _ = trace.agglo_trace(Cs, link)
    lab = trace.labels_at(mg, N, K)
    ref = fcluster(sp_linkage(Cs, link), K, "maxclust") - 1
    rows.append({"linkage": link, "ARI vs final state": round(adjusted_rand(final_state, lab), 3), "matched accuracy": round(trace.match_labels(final_state, lab)["accuracy"], 3), "agrees with scipy (ARI)": round(adjusted_rand(ref, lab), 3)})
pd.DataFrame(rows).set_index("linkage")
````

Comparing linkages against the true labels and then keeping the best one is a mild form of label peeking, so treat the best row as slightly optimistic.

## Step 6 · Hidden Markov model (uses the order of turns)

$$P(z_t\mid z_{t-1}) = A_{z_{t-1}z_t}, \qquad x_t\mid z_t\sim\mathcal N(\mu_{z_t},\mathrm{diag}\,\sigma^2_{z_t})$$

K-means and the GMM look at one turn at a time. The HMM also uses the *sequence*: a refusal tends to be followed by a refusal. It is fitted with `hmmlearn` (EM across all conversations), decodes the most likely state per turn (Viterbi), and the states are named by their best match to the truth.

````python
hm = trace.hmm_trace(X_raw, df["conversation_id"].values, df["timestep_in_conversation"].values, y, seed=SEED)
print(f"accuracy after matching {hm['accuracy']:.4f} | ARI {hm['ari']:.4f} | log-likelihood {hm['log_likelihood']:.1f}")
display(pd.DataFrame(hm["transmat"], index=[f"from {c}" for c in COMPLIANCE_STATES], columns=[f"to {c}" for c in COMPLIANCE_STATES]).round(3))

fig, ax = plt.subplots(figsize=(4.2, 3.4)); ax.imshow(hm["transmat"], cmap="Blues", vmin=0, vmax=1)
for i in range(3):
    for j in range(3): ax.text(j, i, f"{hm['transmat'][i, j]:.2f}", ha="center", va="center", color="#0F172A" if hm["transmat"][i, j] < .6 else "white")
ax.set_xticks(range(3)); ax.set_xticklabels(["ref", "part", "full"]); ax.set_yticks(range(3)); ax.set_yticklabels(["ref", "part", "full"]); ax.set_title("Learned transition matrix"); plt.show()
````

````python
# watch it decode one conversation: scores per turn, decoded state strip, true state underneath
cid = hm["conv_ids"][int(np.argmax(hm["lengths"]))]            # the longest conversation
sel = hm["conv_sorted"] == cid
pred = np.array([hm["match"]["mapping"].get(int(v), -1) for v in hm["decoded_sorted"][sel]]); true = hm["y_sorted"][sel]; Xc = hm["X_sorted"][sel]
turns = np.arange(1, len(Xc) + 1)
fig, ax = plt.subplots(figsize=(8, 3.2))
for j, col in enumerate(SCORE_COLUMNS): ax.plot(turns, Xc[:, j], "o-", color=PALETTE[j], lw=2, label=col.replace("judge_", "p("))
for t_, p_, q_ in zip(turns, pred, true):
    ax.axvspan(t_ - .5, t_ + .5, ymin=0, ymax=.06, color=STATE_COLORS[p_] if p_ >= 0 else "#CBD5E1")
    ax.axvspan(t_ - .5, t_ + .5, ymin=.06, ymax=.12, color=STATE_COLORS[q_], alpha=.5)
ax.set_xticks(turns); ax.set_ylim(0, 1.02); ax.set_xlabel("turn"); ax.set_ylabel("judge score"); ax.legend(fontsize=8, ncol=3, loc="upper center")
ax.set_title(f"{cid}: bottom strip = decoded state, second strip = true state"); plt.show()
print("decoded:", [COMPLIANCE_STATES[p] if p >= 0 else "?" for p in pred]); print("true:   ", [COMPLIANCE_STATES[q] for q in true])
````

## Step 7 · Scoreboard and cross-checks

Only now are the true labels used. **Matched accuracy** names the clusters by the best one-to-one assignment to the true classes (Hungarian algorithm). **ARI** counts pairs of points grouped together in both answers (1 perfect, 0 random). **Silhouette** needs no labels.

Each from-scratch method is also compared with the library version on the same data (`agrees with reference`: 1.0 means the same partition).

````python
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
methods = {
    "K-means (ours, NumPy)": best["labels"],
    "K-means (sklearn)": KMeans(K, n_init=10, random_state=SEED).fit_predict(Z),
    "GMM-EM (ours, NumPy)": gm["labels"],
    "GMM (sklearn)": GaussianMixture(K, covariance_type="full", init_params="kmeans", random_state=SEED).fit(Z).predict(Z),
    "HMM (hmmlearn, uses order)": hm["labels"],
}
refs = {"K-means (ours, NumPy)": "K-means (sklearn)", "GMM-EM (ours, NumPy)": "GMM (sklearn)"}
rows = []
for name, lab in methods.items():
    sc = trace.score(Z, np.asarray(lab), y)
    r = {"method": name, "matched accuracy": round(sc["accuracy"], 4), "ARI": round(sc["ari"], 4), "silhouette": round(sc["silhouette"], 4)}
    if name in refs: r["agrees with reference (ARI)"] = round(adjusted_rand(np.asarray(methods[refs[name]]), np.asarray(lab)), 4)
    rows.append(r)
board = pd.DataFrame(rows).set_index("method"); board
````

````python
# confusion matrix after matching, for the K-means result
m = trace.match_labels(y, best["labels"])
cm = np.array([[np.sum((y == c) & (m["pred"] == p)) for p in m["classes"]] for c in m["classes"]])
fig, ax = plt.subplots(figsize=(4.4, 3.6)); ax.imshow(cm, cmap="Blues")
for i in range(3):
    for j in range(3): ax.text(j, i, cm[i, j], ha="center", va="center", color="#0F172A" if cm[i, j] < cm.max() * .6 else "white")
ax.set_xticks(range(3)); ax.set_xticklabels(["ref", "part", "full"]); ax.set_yticks(range(3)); ax.set_yticklabels(["ref", "part", "full"])
ax.set_xlabel("predicted (after matching)"); ax.set_ylabel("true"); ax.set_title("K-means confusion matrix"); plt.show()
print("cluster -> state mapping found by the Hungarian algorithm:", {int(c): COMPLIANCE_STATES[s] for c, s in m["mapping"].items()})
````

````python
# automatic sanity checks: the from-scratch code must agree with the library versions
assert adjusted_rand(methods["K-means (sklearn)"], methods["K-means (ours, NumPy)"]) > 0.97, "K-means disagrees with sklearn"
assert abs(trace.score(Z, methods["K-means (ours, NumPy)"], y)["accuracy"] - trace.score(Z, methods["K-means (sklearn)"], y)["accuracy"]) < 0.02
assert all(b >= a - 1e-8 for a, b in zip(gm["lls"], gm["lls"][1:])), "EM log-likelihood decreased"
assert np.allclose(hm["transmat"].sum(1), 1)
assert abs(pca["ratio"][-1]) < 1e-9, "scores should sum to 1, so the last eigenvalue is ~0"
print("all cross-checks passed")
````

## What to read from this

* Compare the scoreboard across campaigns by changing `CAMPAIGN` at the top: the **noisy** judge should hurt every method, the **stealthy** campaign changes how rare full violation is, and **persistent** gives the HMM a real sequence to model.
* Labels are only used in the scoring cells above.
* Everything is simulated. The honest test is the real data (notebook 01, section 6, and the UI).
